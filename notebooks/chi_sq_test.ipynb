```python
import numpy as np
import pandas as pd

import scipy.stats as stats

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('../data/boat_vs_noise_audio_products.csv')
```

```python
df[['brand', 'has_anc']].info()

print(df['brand'].value_counts(dropna=False))
print(df['has_anc'].value_counts(dropna=False))

pd.crosstab(df['brand'], df['has_anc'])
```

```python
import scipy.stats as stats

observed = pd.crosstab(df['brand'], df['has_anc'])

chi2, p_value, dof, expected = stats.chi2_contingency(
    observed,
    correction=False
)

print("Observed:\n", observed)
print("\nExpected:\n", expected)
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("P-value:", p_value)
```

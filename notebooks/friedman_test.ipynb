```python
import numpy as np
import pandas as pd
from scipy.stats import friedmanchisquare
```

```python
df = pd.DataFrame({
    "Method_A": [50, 60, 55, 65, 70, 58, 62, 48, 75, 68, 52, 64],
    "Method_B": [55, 65, 57, 68, 72, 62, 60, 55, 78, 70, 58, 67],
    "Method_C": [60, 63, 62, 72, 75, 66, 68, 58, 80, 74, 61, 70]
}, index=[f"Student_{i}" for i in range(1, 13)])

df.index.name = "Student"

df
```

```python
print("Missing values:")
print(df.isnull().sum())

print("\nNumber of students:", df.shape[0])

print("Number of methods:", df.shape[1])

assert df.notna().all().all(), "Missing observations found!"
assert df.shape[1] >= 3, "Friedman test requires at least 3 conditions."
```

```python
stat, p_value = friedmanchisquare(
    df["Method_A"],
    df["Method_B"],
    df["Method_C"]
)

print("Friedman Statistic:", stat)
print("P-value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0: Significant difference among teaching methods.")
else:
    print("Fail to reject H0: Insufficient evidence of a difference.")
```

```python
n = df.shape[0]
k = df.shape[1]

effect_size = stat / (n * (k - 1))

print("Kendall's W:", effect_size)
```

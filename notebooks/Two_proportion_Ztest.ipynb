```python

import numpy as np
from scipy.stats import norm
```

```python
# Website A
x1 = 60      # Number of conversions (successes)
n1 = 400     # Total visitors

# Website B
x2 = 100     # Number of conversions (successes)
n2 = 500     # Total visitors

alpha = 0.05  # Significance level (5%)
```

```python
p1 = x1 / n1
p2 = x2 / n2

print("Website A conversion rate:", p1)
print("Website B conversion rate:", p2)
```

```python
p_pool = (x1 + x2) / (n1 + n2)

print("Pooled proportion:", p_pool)
print("Pooled conversion rate:", p_pool * 100, "%")
```

```python
# Expected counts for Website A
expected_success_A = n1 * p_pool
expected_failure_A = n1 * (1 - p_pool)

# Expected counts for Website B
expected_success_B = n2 * p_pool
expected_failure_B = n2 * (1 - p_pool)

print("Website A:")
print("Expected successes:", expected_success_A)
print("Expected failures:", expected_failure_A)

print("\nWebsite B:")
print("Expected successes:", expected_success_B)
print("Expected failures:", expected_failure_B)
```

```python
from statsmodels.stats.proportion import proportions_ztest

count = np.array([x1, x2])
nobs = np.array([n1, n2])

z_stat, p_value = proportions_ztest(
    count=count,
    nobs=nobs,
    alternative="two-sided"
)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

```python
if p_value < alpha:
    print("Reject the null hypothesis.")
    print("The proportions are statistically significantly different.")
else:
    print("Fail to reject the null hypothesis.")
    print("Insufficient evidence of a significant difference.")
```

```python
print(f"Website A conversion rate: {p1:.2%}")
print(f"Website B conversion rate: {p2:.2%}")
print(f"Observed difference (A - B): {p1 - p2:.2%}")
print(f"Z-statistic: {z_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
from statsmodels.stats.proportion import proportion_effectsize

# Cohen's h
effect_size = proportion_effectsize(p1, p2)

# Raw difference between sample proportions
observed_difference = p1 - p2

print("Observed difference:", observed_difference)
print("Cohen's h:", effect_size)
print("Absolute effect size:", abs(effect_size))
```

```python
from statsmodels.stats.power import NormalIndPower

power_analysis = NormalIndPower()

required_n1 = power_analysis.solve_power(
    effect_size=abs(effect_size),
    power=0.80,
    alpha=0.05,
    ratio=n2 / n1,
    alternative="two-sided"
)

required_n2 = required_n1 * (n2 / n1)

print("Desired power: 80%")
print("Significance level:", alpha)
print("Required sample size for Website A:", int(np.ceil(required_n1)))
print("Required sample size for Website B:", int(np.ceil(required_n2)))
print("Total required sample size:",
      int(np.ceil(required_n1)) + int(np.ceil(required_n2)))
```

```python
from statsmodels.stats.proportion import confint_proportions_2indep

ci_low, ci_high = confint_proportions_2indep(
    count1=x1,
    nobs1=n1,
    count2=x2,
    nobs2=n2,
    compare="diff",
    method="wald",
    alpha=0.05
)

print("Observed difference (A - B):", p1 - p2)
print("95% Confidence Interval:", (ci_low, ci_high))
print(f"Lower bound: {ci_low:.4f} ({ci_low:.2%})")
print(f"Upper bound: {ci_high:.4f} ({ci_high:.2%})")
```
